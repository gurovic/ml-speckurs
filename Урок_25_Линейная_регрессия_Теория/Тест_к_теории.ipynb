# Занятие 25. Мини-задачки: линейная регрессия

Короткая проверка понимания теории из `Урок_25_Линейная_регрессия.ipynb`.

Все задания приведены к формату для LMS. У каждого задания явно указан тип, варианты ответа там, где они нужны, и правильный ответ.


## 1. Что именно делает модель линейной

**Тип:** выбор ответа

**Задание.** Ученик добавил в таблицу признаки `x`, `x^2`, `x^3` и `x1 * x2`, а затем обучил `LinearRegression`. Он говорит: «Это уже не линейная регрессия, потому что график может быть кривой». Какой ответ точнее всего исправляет это рассуждение?

**Варианты ответа:**

A. Ученик прав: если среди признаков есть степени, `LinearRegression` перестаёт быть линейной моделью.  
B. Модель линейна только тогда, когда в ней ровно один признак и один коэффициент.  
C. Модель остаётся линейной по коэффициентам: она складывает подготовленные признаки с весами, даже если сами признаки устроены сложно.  
D. Модель становится линейной только после масштабирования всех признаков.

**Правильный ответ:** C.


## 2. Когда коэффициент нельзя читать как причинный эффект

**Тип:** выбор ответа

**Задание.** В модели цены квартиры коэффициент при `площадь_м2` положительный. В данных большие квартиры чаще находятся в новых домах и дорогих районах, а район и год постройки в модель не попали. Какой вывод самый корректный?

**Варианты ответа:**

A. Коэффициент показывает связь в построенной модели, но его нельзя автоматически считать чистым причинным эффектом площади.  
B. Коэффициент доказывает, что увеличение площади всегда само по себе вызывает ровно такой рост цены.  
C. Если коэффициент положительный, остальные признаки и устройство данных уже не важны.  
D. Линейная регрессия запрещает интерпретировать коэффициенты при любых обстоятельствах.

**Правильный ответ:** A.


## 3. Выбор метрики при редких больших ошибках

**Тип:** выбор ответа

**Задание.** Две модели имеют похожую обычную ошибку на большинстве квартир, но первая иногда ошибается на очень дорогих объектах на десятки миллионов. Для бизнеса такие редкие крупные промахи особенно опасны. Какая метрика лучше отразит эту проблему?

**Варианты ответа:**

A. Только $R^2$, потому что он всегда измеряет ошибку в рублях.  
B. MAE, потому что она сильнее квадратичных метрик штрафует крупные промахи.  
C. Accuracy, потому что цена — это числовой ответ.  
D. MSE или RMSE, потому что квадрат ошибки резко усиливает вклад больших промахов.

**Правильный ответ:** D.


## 4. Отрицательный $R^2$ на новых данных

**Тип:** выбор ответа

**Задание.** На validation модель получила $R^2=-0.3`. Обучающийся считает, что это невозможно, потому что «$R^2$ — это процент объяснённой ошибки, он не бывает меньше нуля». Что на самом деле означает такой результат?

**Варианты ответа:**

A. В коде обязательно перепутаны признаки и целевая переменная.  
B. На этих данных модель хуже константного baseline, который предсказывает среднее.  
C. Модель идеально переобучилась на validation.  
D. Значит, MSE модели равна нулю.

**Правильный ответ:** B.


## 5. Утечка при масштабировании

**Тип:** выбор ответа

**Задание.** Ученик сначала применил `StandardScaler.fit_transform()` ко всей таблице признаков, а потом разделил данные на train и test. Почему это плохая схема проверки качества?

**Варианты ответа:**

A. Масштабирование вообще нельзя использовать вместе с линейной регрессией.  
B. После масштабирования признаки перестают быть числовыми.  
C. В параметры масштабирования попала информация из test, поэтому проверка стала менее честной.  
D. `fit_transform()` всегда удаляет целевую переменную.

**Правильный ответ:** C.


## 6. Регуляризация и выбор `alpha`

**Тип:** выбор ответа

**Задание.** Ridge-модель с полиномиальными признаками сильно переобучается. Ученик увеличивает `alpha` и видит, что train-ошибка растёт, а validation-ошибка сначала падает, потом снова растёт. Какой вывод наиболее правильный?

**Варианты ответа:**

A. Любое увеличение `alpha` всегда улучшает модель на новых данных.  
B. `alpha` нужно выбирать по validation или кросс-валидации: слишком слабый штраф не спасает от переобучения, слишком сильный ведёт к недообучению.  
C. Если train-ошибка растёт, модель обязательно становится хуже по всем критериям.  
D. Регуляризация меняет только свободный член `b`, а не коэффициенты признаков.

**Правильный ответ:** B.


## 7. Мультиколлинеарность и качество прогноза

**Тип:** выбор ответа

**Задание.** В таблице есть `площадь_м2`, `площадь_см2` и `площадь_квартиры_округленная`. После небольшого изменения train коэффициенты при этих признаках сильно меняются, но качество прогноза почти не меняется. Как это лучше интерпретировать?

**Варианты ответа:**

A. Если качество не меняется, коэффициенты точно можно интерпретировать как устойчивые эффекты каждого признака.  
B. Это доказывает, что все три признака обязательно нужно оставить без изменений.  
C. Это означает, что линейная регрессия перестала минимизировать MSE.  
D. Это типичный признак мультиколлинеарности: модель может хорошо предсказывать, но плохо «делит вклад» между похожими признаками.

**Правильный ответ:** D.


## 8. Остатки как диагностика формы зависимости

**Тип:** выбор ответа

**Задание.** На графике `x` против остатков видна устойчивая дуга: в середине модель в основном завышает прогноз, а по краям занижает. Средняя ошибка почти равна нулю. Что это говорит о модели?

**Варианты ответа:**

A. Почти нулевая средняя ошибка доказывает, что модель хорошая.  
B. Нужно оценивать только свободный член, потому что коэффициент при `x` не влияет на остатки.  
C. Остатки показывают пропущенную нелинейную закономерность; стоит проверить сложные признаки или другую модель.  
D. Дуга на остатках означает, что test использовался строго честно.

**Правильный ответ:** C.


## 9. Экстраполяция за пределы данных

**Тип:** выбор ответа

**Задание.** Модель обучалась на квартирах площадью от 25 до 120 м². Для объекта площадью 400 м² она выдаёт очень высокую цену, потому что продолжает найденную линейную зависимость. Какой вывод самый осторожный?

**Варианты ответа:**

A. Такой прогноз требует осторожности: модель применили далеко за пределами области, на которой она училась.  
B. Прогнозу можно доверять так же, как прогнозам внутри диапазона train: линейная регрессия всегда корректно экстраполирует.  
C. Если площадь больше всех train-значений, модель обязана вернуть ошибку вместо прогноза.  
D. Экстраполяция опасна только для классификации, но не для регрессии.

**Правильный ответ:** A.


## 10. Test после множества попыток

**Тип:** выбор ответа

**Задание.** Команда сравнила 30 вариантов признаков, разные степени полиномов и несколько значений регуляризации. Каждый вариант оценивали на test, а в отчёт записали лучший результат на test. Что принципиально не так?

**Варианты ответа:**

A. Ничего: чем больше раз посмотреть на test, тем точнее итоговая оценка.  
B. На test разрешено выбирать признаки, но запрещено выбирать только `alpha`.  
C. Проблема была бы только при классификации, а для регрессии так делать можно.  
D. Test превратился в инструмент выбора модели; итоговая оценка стала оптимистичной и уже не является независимой проверкой.

**Правильный ответ:** D.
