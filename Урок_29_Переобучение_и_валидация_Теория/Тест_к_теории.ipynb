# Занятие 29. Мини-задачки: переобучение и валидация

Короткая проверка понимания теории из `Урок_29_Переобучение_и_валидация.ipynb`.

Все задания приведены к формату для LMS. У каждого задания явно указан тип, варианты ответа там, где они нужны, и правильный ответ.


## 1. Диагноз по ошибкам

**Тип:** выбор ответа

**Задание.** У модели очень низкая ошибка на train и высокая ошибка на validation. Что это чаще всего означает?

**Варианты ответа:**

A. Модель не видела train.  
B. Validation нельзя использовать для оценки.  
C. Модель переобучилась под train.  
D. Test автоматически испортился.

**Правильный ответ:** C.


## 2. Недообучение

**Тип:** выбор ответа

**Задание.** У модели высокая ошибка и на train, и на validation. Какой вывод наиболее вероятен?

**Варианты ответа:**

A. Модель идеально обобщается.  
B. Test использовался слишком часто.  
C. Нужно удалить validation.
D. Модель слишком простая или признаков недостаточно.  

**Правильный ответ:** D.


## 3. Почему test нельзя открывать много раз

**Тип:** открытый ответ

**Задание.** Почему нельзя много раз подбирать модель по результату на test, даже если test не используется в `fit`?

**Формат ответа:** короткое текстовое объяснение.

**Правильный ответ:** потому что решения начинают подстраиваться под test, и финальная оценка становится оптимистичной и уже не независимой.


## 4. Validation curve

**Тип:** выбор ответа

**Задание.** На validation curve при увеличении глубины дерева train-качество улучшается, а validation-качество сначала улучшается, потом ухудшается. Как выбирать глубину?

**Варианты ответа:**

A. Всегда брать максимальную глубину.  
B. Брать глубину с лучшим validation-качеством, а не с лучшим train-качеством.  
C. Брать минимальную глубину независимо от качества.  
D. Выбирать глубину по test после каждого запуска.

**Правильный ответ:** B.


## 5. K-fold CV

**Тип:** открытый ответ

**Задание.** Зачем использовать k-fold cross-validation вместо одного случайного validation split?

**Формат ответа:** короткое текстовое объяснение.

**Правильный ответ:** чтобы оценка меньше зависела от одного случайного разбиения и каждая часть данных побывала в validation.


## 6. GroupKFold

**Тип:** выбор ответа

**Задание.** В данных много строк на одного пользователя. Нужно оценить качество на новых пользователях. Какое разбиение лучше использовать?

**Варианты ответа:**

A. GroupKFold или group split по идентификатору пользователя.  
B. Случайно перемешать строки одного пользователя между train и validation.  
C. Обучаться и проверяться на одних и тех же строках.  
D. Использовать test для подбора групп.

**Правильный ответ:** A.


## 7. Time split

**Тип:** выбор ответа

**Задание.** Модель должна прогнозировать продажи в будущем по историческим данным. Какое validation-разбиение наиболее честное?

**Варианты ответа:**

A. Случайно перемешать все даты, включая будущие.  
B. Обучаться на будущем, проверяться на прошлом.  
C. Обучаться на более ранних датах, проверяться на более поздних.  
D. Удалить дату из всех данных и не проверять качество.

**Правильный ответ:** C.


## 8. Preprocessing внутри fold

**Тип:** выбор ответа

**Задание.** Почему scaler или encoder при cross-validation нужно обучать внутри каждого fold только на train-части fold?

**Варианты ответа:**

A. Чтобы модель не могла обучиться.  
B. Чтобы все признаки стали текстовыми.  
C. Чтобы validation-часть стала больше train-части.
D. Чтобы preprocessing не видел validation-часть fold и не давал утечку.  

**Правильный ответ:** D.


## 9. Learning curve

**Тип:** выбор ответа

**Задание.** На learning curve train и validation ошибки близки друг к другу и обе остаются высокими даже при добавлении данных. Что это подсказывает?

**Варианты ответа:**

A. Нужно меньше данных, потому что данных уже слишком много.  
B. Модель, вероятно, недообучается; стоит усложнить модель или признаки.  
C. Модель точно переобучилась.  
D. Test нужно использовать для обучения.

**Правильный ответ:** B.
