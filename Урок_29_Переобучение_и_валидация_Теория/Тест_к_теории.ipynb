# Занятие 29. Мини-задачки: переобучение и валидация

Короткая проверка понимания теории из основного ноутбука занятия.

Все задания приведены к формату для LMS. У каждого задания явно указан тип, варианты ответа и правильный ответ.


## 1. Train-качество не доказывает обобщение

**Тип:** выбор ответа

**Задание.** Модель получила почти нулевую ошибку на train после добавления большого числа признаков, но validation-ошибка заметно выросла. Какой вывод самый разумный?

**Варианты ответа:**

A. Модель стала лучше, потому что train-ошибка — главный показатель.  
B. Похоже на переобучение: модель подстроилась под train и хуже переносится на новые данные.  
C. Validation нужно удалить, потому что он мешает обучению.  
D. Нулевая train-ошибка невозможна в регрессии.  

**Правильный ответ:** B.


## 2. Validation и test выполняют разные роли

**Тип:** выбор ответа

**Задание.** Ученик выбирал степень полинома по test, потому что “test самый честный”. Почему это ломает смысл проверки?

**Варианты ответа:**

A. Test всегда должен быть больше train.  
B. По test нельзя считать метрики регрессии.  
C. Полиномиальные признаки можно выбирать только по train-ошибке.  
D. Test начинает участвовать в выборе модели и перестаёт быть независимой итоговой проверкой.  

**Правильный ответ:** D.


## 3. Validation curve: что искать

**Тип:** выбор ответа

**Задание.** На validation curve при росте `max_depth` train-качество всё улучшается, а validation сначала улучшается, затем ухудшается. Как выбирать `max_depth`?

**Варианты ответа:**

A. Выбрать область лучшего validation-качества, учитывая разброс и простоту модели.  
B. Взять максимальную глубину: train-качество там лучше.  
C. Взять минимальную глубину: она всегда самая устойчивая.  
D. Выбрать глубину по test после построения графика.  

**Правильный ответ:** A.


## 4. Learning curve и нехватка данных

**Тип:** выбор ответа

**Задание.** Learning curve показывает: train-качество высокое, validation заметно хуже, но при добавлении данных разрыв сокращается. Что это может означать?

**Варианты ответа:**

A. Добавление данных всегда ухудшит validation.  
B. Это доказывает, что модель недообучена.  
C. Модель, вероятно, выигрывает от дополнительных данных: разброс уменьшается.  
D. Learning curve нельзя строить для sklearn-моделей.  

**Правильный ответ:** C.


## 5. Когда обычный KFold не подходит

**Тип:** выбор ответа

**Задание.** Есть данные по нескольким измерениям одного и того же ученика. Если случайно разложить строки по fold, один ученик может попасть и в train, и в validation. Что лучше использовать?

**Варианты ответа:**

A. Обычный KFold, потому что он всегда самый честный.  
B. StratifiedKFold, потому что он группирует строки одного ученика.  
C. TimeSeriesSplit, даже если времени в данных нет.  
D. GroupKFold или другое групповое разбиение, чтобы один ученик не попадал в разные части проверки.  

**Правильный ответ:** D.


## 6. TimeSeriesSplit против перемешивания

**Тип:** выбор ответа

**Задание.** Модель прогнозирует спрос на следующую неделю. Ученик случайно перемешал даты и сделал обычный train/test split. В чём риск?

**Варианты ответа:**

A. Случайное перемешивание всегда запрещено для любых данных.  
B. В train может попасть будущее относительно validation/test, и оценка станет слишком оптимистичной.  
C. Временные признаки перестанут быть числовыми.  
D. Модель не сможет вызвать `.predict()`.  

**Правильный ответ:** B.


## 7. Pipeline внутри CV

**Тип:** выбор ответа

**Задание.** Почему при кросс-валидации предобработку лучше включать в `Pipeline`, а не делать `fit_transform` один раз до CV?

**Варианты ответа:**

A. Pipeline гарантирует, что scaler/imputer/encoder обучаются только на train-части каждого fold.  
B. Pipeline ускоряет код, но на честность проверки не влияет.  
C. Без Pipeline нельзя импортировать метрики.  
D. Pipeline нужен только для нейронных сетей.  

**Правильный ответ:** A.


## 8. GridSearchCV и внешний test

**Тип:** выбор ответа

**Задание.** После `GridSearchCV` ученик сообщает лучший score из CV как итоговое качество проекта. Что важно добавить для честного финального вывода?

**Варианты ответа:**

A. Переобучить модель отдельно на каждом объекте test.  
B. Выбрать параметры повторно по test.  
C. Проверить уже выбранную схему на отложенном test, который не участвовал в подборе.  
D. Считать только train-score лучшей модели.  

**Правильный ответ:** C.


## 9. Фиксация random_state

**Тип:** выбор ответа

**Задание.** Один и тот же ноутбук даёт разные метрики при каждом запуске, потому что split и модели со случайностью не фиксированы. Почему это проблема?

**Варианты ответа:**

A. Случайность запрещена во всех алгоритмах ML.  
B. Без фиксированных seed трудно воспроизвести эксперимент и понять, что реально улучшило качество.  
C. `random_state` меняет целевую переменную.  
D. Фиксировать seed нужно только в финальной ячейке, после всех экспериментов.  

**Правильный ответ:** B.


## 10. Много экспериментов и “победа случайности”

**Тип:** выбор ответа

**Задание.** Ученик проверил 200 вариантов признаков на одном validation и выбрал лучший с небольшим отрывом. Почему к такому улучшению стоит относиться осторожно?

**Варианты ответа:**

A. Validation нельзя использовать для выбора признаков.  
B. Любой отрыв на validation гарантирует такое же улучшение на новых данных.  
C. Число экспериментов не влияет на риск переоценки качества.  
D. Если вариантов много, часть улучшения может быть случайной подгонкой под конкретный validation; полезно перепроверить схему CV или финальным test.  

**Правильный ответ:** D.
