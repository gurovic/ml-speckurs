# Занятие 35. Мини-задачки: градиентный бустинг

Короткая проверка понимания теории из `Урок_35_Boosting.ipynb`.

Все задания приведены к формату для LMS. У каждого задания явно указан тип, варианты ответа там, где они нужны, и правильный ответ.


## 1. Последовательный ансамбль

**Тип:** выбор ответа

**Задание.** В чем главная идея градиентного бустинга?

**Варианты ответа:**

A. Обучить много независимых моделей и просто усреднить их.  
B. Использовать только один ближайший сосед.  
C. Последовательно добавлять слабые модели, каждая из которых исправляет текущие ошибки ансамбля.  
D. Запретить функции потерь.

**Правильный ответ:** C.


## 2. Остатки в регрессии

**Тип:** открытый ответ

**Задание.** Текущий прогноз ансамбля для объекта равен `12`, а правильный ответ равен `15`. Какой остаток должна попытаться объяснить следующая модель, если используем идею `y - ŷ`?

**Формат ответа:** число.

**Правильный ответ:** `3`.


## 3. Learning rate

**Тип:** открытый ответ

**Задание.** Зачем в бустинге нужен learning rate?

**Формат ответа:** короткое текстовое объяснение.

**Правильный ответ:** он уменьшает вклад каждого нового дерева, делая обучение более осторожным и снижая риск переобучения.


## 4. Маленький learning rate

**Тип:** выбор ответа

**Задание.** Что обычно нужно сделать, если сильно уменьшить learning rate и хочется сохранить качество?

**Варианты ответа:**

A. Уменьшить число деревьев до одного.  
B. Удалить validation.  
C. Перестать считать функцию потерь.
D. Увеличить число деревьев и подбирать его по validation.  

**Правильный ответ:** D.


## 5. Сложность базовых деревьев

**Тип:** выбор ответа

**Задание.** Почему в градиентном бустинге часто используют неглубокие деревья?

**Варианты ответа:**

A. Потому что глубокие деревья не умеют делать предсказания.  
B. Потому что слабые базовые модели снижают риск слишком быстрого запоминания шума.  
C. Потому что бустинг работает только без признаков.  
D. Потому что неглубокое дерево всегда идеально на train.

**Правильный ответ:** B.


## 6. Early stopping

**Тип:** открытый ответ

**Задание.** Что означает early stopping в обучении бустинга?

**Формат ответа:** короткое текстовое объяснение.

**Правильный ответ:** обучение останавливают, когда качество на validation перестает улучшаться, чтобы не добавлять лишние деревья.


## 7. Функция потерь

**Тип:** сопоставление

**Задание.** Сопоставьте задачу с типичной функцией потерь.

**Варианты для сопоставления:**

A. Регрессия с MSE  
B. Бинарная классификация  
C. Регрессия с MAE

1. logloss;  
2. квадратичная ошибка;  
3. абсолютная ошибка.

**Правильный ответ:** A-2, B-1, C-3.


## 8. Масштабирование

**Тип:** выбор ответа

**Задание.** Нужно ли обычно масштабировать числовые признаки специально для бустинга на деревьях?

**Варианты ответа:**

A. Обычно нет, потому что деревья работают через пороговые разбиения.  
B. Да, иначе дерево не сможет выбрать порог.  
C. Да, иначе нельзя посчитать logloss.  
D. Масштабирование заменяет обучение модели.

**Правильный ответ:** A.


## 9. Сравнение с Random Forest

**Тип:** сопоставление

**Задание.** Сопоставьте модель с характерным свойством.

**Варианты для сопоставления:**

A. Random Forest  
B. Gradient Boosting

1. последовательное добавление деревьев, исправляющих ошибки;
