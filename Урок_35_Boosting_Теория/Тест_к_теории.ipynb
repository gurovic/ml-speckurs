# Занятие 35. Мини-задачки: градиентный бустинг

Короткая проверка понимания теории из основного ноутбука занятия.

Все задания приведены к формату для LMS. У каждого задания явно указан тип, варианты ответа и правильный ответ.


## 1. Последовательность важнее количества

**Тип:** выбор ответа

**Задание.** Ученик говорит: “градиентный бустинг — это просто много деревьев, как случайный лес”. Что принципиально отличает boosting?

**Варианты ответа:**

A. Все деревья обучаются независимо на bootstrap-выборках.  
B. Деревья обучаются последовательно: новая модель исправляет ошибки текущего ансамбля.  
C. Boosting не использует функцию потерь.  
D. Boosting всегда строит одно большое дерево.  

**Правильный ответ:** B.


## 2. Learning rate и число деревьев

**Тип:** выбор ответа

**Задание.** Модель с маленьким `learning_rate` обучается хуже при том же числе деревьев. Какой вывод корректен?

**Варианты ответа:**

A. Маленький `learning_rate` всегда плох.  
B. `learning_rate` влияет только на масштабирование признаков.  
C. Нужно заменить validation на train.  
D. При меньшем шаге часто нужно больше деревьев; важен совместный подбор `learning_rate` и `n_estimators`.  

**Правильный ответ:** D.


## 3. Слишком много деревьев

**Тип:** выбор ответа

**Задание.** При добавлении деревьев train loss продолжает падать, а validation loss после некоторого момента растёт. Что это означает?

**Варианты ответа:**

A. Бустинг не умеет переобучаться.  
B. Нужно обязательно добавить ещё деревьев.  
C. Ансамбль начал переобучаться; стоит использовать early stopping или выбрать меньше итераций.  
D. Validation loss не имеет смысла для бустинга.  

**Правильный ответ:** C.


## 4. Early stopping и validation

**Тип:** выбор ответа

**Задание.** Почему validation-набор для early stopping нельзя одновременно считать полноценным финальным test?

**Варианты ответа:**

A. Он участвовал в выборе момента остановки, значит уже повлиял на модельное решение.  
B. На нём нельзя считать метрики.  
C. Early stopping не использует validation.  
D. Test должен обязательно быть меньше validation.  

**Правильный ответ:** A.


## 5. Слабые деревья в бустинге

**Тип:** выбор ответа

**Задание.** Почему в градиентном бустинге часто используют неглубокие деревья, а не очень глубокие?

**Варианты ответа:**

A. Неглубокие деревья быстрее запоминают каждый объект.  
B. Глубокие деревья нельзя складывать.  
C. Неглубокие деревья запрещают регуляризацию.  
D. Слабые деревья дают контролируемые маленькие исправления; глубокие могут слишком быстро подстроиться под шум.  

**Правильный ответ:** D.


## 6. Бустинг чувствителен к шуму

**Тип:** выбор ответа

**Задание.** В целевой переменной есть заметный шум и ошибки разметки. Почему бустинг может начать ухудшаться при долгом обучении?

**Варианты ответа:**

A. Он вообще не видит целевую переменную.  
B. Он последовательно исправляет остатки и может начать подстраиваться под шумные ошибки.  
C. Он усредняет независимые модели и поэтому не может реагировать на шум.  
D. Он автоматически удаляет все выбросы.  

**Правильный ответ:** B.


## 7. HistGradientBoosting и масштабирование

**Тип:** выбор ответа

**Задание.** Для табличных числовых признаков ученик хочет обязательно применять StandardScaler перед HistGradientBoostingRegressor “как перед Ridge”. Что лучше сказать?

**Варианты ответа:**

A. Без StandardScaler бустинг не запускается.  
B. Масштабирование превращает бустинг в линейную регрессию.  
C. Деревьям и бустингу на деревьях масштабирование обычно не так критично, как линейным моделям с регуляризацией.  
D. Scaler нужно обучать на test, чтобы качество было честным.  

**Правильный ответ:** C.


## 8. Сравнение библиотек

**Тип:** выбор ответа

**Задание.** XGBoost, LightGBM и CatBoost дали разное качество и разное время обучения. Как корректнее выбирать финальную модель?

**Варианты ответа:**

A. Сравнить на одной честной схеме validation/CV, учитывая метрику, время, стабильность и ограничения данных.  
B. Взять самую известную библиотеку независимо от validation.  
C. Взять модель с лучшим train-score.  
D. Выбрать ту, где больше параметров.  

**Правильный ответ:** A.


## 9. Категориальные признаки

**Тип:** выбор ответа

**Задание.** Почему CatBoost часто удобен на задачах с категориальными признаками?

**Варианты ответа:**

## 10. Importance в бустинге

**Тип:** выбор ответа

**Задание.** У модели высокая важность признака `user_id`. Почему это подозрительно?

**Варианты ответа:**

A. Любой id-признак всегда лучший и безопасный.  
B. Это может быть признак утечки или запоминания отдельных пользователей; важность нужно интерпретировать вместе с постановкой задачи и проверкой на новых данных.  
C. Бустинг не умеет использовать id-признаки.  
D. Если importance высокая, feature leakage невозможна.  

**Правильный ответ:** B.
