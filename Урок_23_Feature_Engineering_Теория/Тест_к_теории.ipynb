# Занятие 23. Мини-задачки: конструирование признаков

Короткая проверка понимания теории из `Урок_23_Feature_Engineering.ipynb`.

Все задания приведены к формату для LMS. У каждого задания явно указан тип, варианты ответа там, где они нужны, и правильный ответ.


## 1. Признак без утечки

**Тип:** выбор ответа

**Задание.** Модель должна предсказывать отток пользователя на начало месяца. Какой признак безопаснее всего использовать?

**Варианты ответа:**

A. Был ли пользователь возвращен скидкой в конце месяца.  
B. Факт удаления аккаунта через две недели после прогноза.  
C. Число обращений в поддержку за прошлый месяц.  
D. Итоговый статус пользователя в конце месяца.

**Правильный ответ:** C.


## 2. Нормировка на размер

**Тип:** открытый ответ

**Задание.** У одного магазина 100 заказов и 5 возвратов, у другого 10 заказов и 2 возврата. Почему признак `число_возвратов` может быть хуже, чем `доля_возвратов`?

**Формат ответа:** короткое текстовое объяснение.

**Правильный ответ:** абсолютное число возвратов зависит от размера магазина, а доля возвратов лучше показывает риск возврата на один заказ.


## 3. Логарифмирование признака

**Тип:** выбор ответа

**Задание.** Доход пользователей имеет длинный правый хвост: большинство значений небольшие, но есть очень крупные. Зачем можно добавить признак `log(доход + 1)`?

**Варианты ответа:**

A. Чтобы сделать все доходы одинаковыми.  
B. Чтобы удалить связь дохода с целевой переменной.  
C. Чтобы превратить числовой признак в категориальный.
D. Чтобы сжать крупные значения и уменьшить влияние экстремального масштаба.  

**Правильный ответ:** D.


## 4. Признак-взаимодействие

**Тип:** открытый ответ

**Задание.** Цена доставки зависит не только от веса и расстояния по отдельности, но и от их совместного эффекта. Какой простой признак-взаимодействие можно добавить?

**Формат ответа:** формула признака.

**Правильный ответ:** `вес * расстояние`.


## 5. One-hot или ordinal encoding

**Тип:** сопоставление

**Задание.** Сопоставьте тип категориального признака с подходящим кодированием.

**Варианты для сопоставления:**

A. Цвет товара: красный, зеленый, синий  
B. Уровень образования: школа, бакалавриат, магистратура  
C. День недели

1. можно рассмотреть порядковое кодирование, если порядок действительно важен;  
2. чаще подходит one-hot, потому что нет естественного порядка;  
3. можно использовать циклические признаки или one-hot, потому что категории повторяются по кругу.

**Правильный ответ:** A-2, B-1, C-3.


## 6. Неизвестная категория

**Тип:** выбор ответа

**Задание.** В validation появился город, которого не было в train. Какой вариант обработки категорий наиболее устойчив к такой ситуации?

**Варианты ответа:**

A. Удалить весь validation.  
B. Настроить encoder так, чтобы неизвестные категории попадали в отдельную или нулевую кодировку.  
C. Обучить encoder на train и validation вместе.  
D. Заменить целевую переменную на название города.

**Правильный ответ:** B.


## 7. Агрегаты без утечки

**Тип:** выбор ответа

**Задание.** Для прогноза покупки в марте вы хотите добавить `средний_чек_пользователя`. Как посчитать этот признак без утечки?

**Варианты ответа:**

A. Только по покупкам, которые были доступны до момента прогноза.  
B. По всем покупкам пользователя, включая март и будущие месяцы.  
C. По целевой переменной validation.  
D. Отдельно по test после просмотра правильных ответов.

**Правильный ответ:** A.


## 8. Проверка полезности признака

**Тип:** открытый ответ

**Задание.** Вы добавили новый признак, и качество на train выросло, а на validation ухудшилось. Какой вывод нельзя делать автоматически и что нужно проверить?

**Формат ответа:** короткое текстовое объяснение.

**Правильный ответ:** нельзя считать признак полезным только по train; нужно проверить переобучение, утечку и качество на validation.
