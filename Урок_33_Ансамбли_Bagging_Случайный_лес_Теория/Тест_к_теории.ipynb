# Занятие 33. Мини-задачки: bagging и случайный лес

Короткая проверка понимания теории из основного ноутбука занятия.

Все задания приведены к формату для LMS. У каждого задания явно указан тип, варианты ответа и правильный ответ.


## 1. Зачем bagging нестабильным моделям

**Тип:** выбор ответа

**Задание.** Одиночное дерево сильно меняется при небольшом изменении train. Почему bagging может помочь?

**Варианты ответа:**

A. Он обучает несколько моделей на bootstrap-выборках и усредняет/голосует, снижая разброс предсказаний.  
B. Он делает одно дерево глубже.  
C. Он заменяет все признаки случайным шумом.  
D. Он гарантирует отсутствие bias.  

**Правильный ответ:** A.


## 2. Bootstrap не означает новые данные

**Тип:** выбор ответа

**Задание.** В bootstrap-выборке некоторые объекты повторяются, а некоторые не попадают вообще. Как это понимать?

**Варианты ответа:**

A. Bootstrap создаёт новые реальные наблюдения из будущего.  
B. Bootstrap можно делать только для test.  
C. Это выборка с возвращением из train: она меняет состав обучающих наборов для базовых моделей.  
D. Повторы объектов делают обучение невозможным.  

**Правильный ответ:** C.


## 3. OOB не заменяет всё

**Тип:** выбор ответа

**Задание.** RandomForest обучен с `oob_score=True`, и OOB-качество выглядит хорошим. Как правильно использовать этот результат?

**Варианты ответа:**

A. OOB — это качество на скрытом test из платформы.  
B. Если есть OOB, validation и test всегда запрещены.  
C. OOB считается только для линейных моделей.  
D. OOB можно использовать как дополнительную внутреннюю оценку, но финальный test всё равно нужен для итогового вывода.  

**Правильный ответ:** D.


## 4. Случайные признаки в лесу

**Тип:** выбор ответа

**Задание.** Зачем RandomForest на каждом split рассматривает не все признаки, а случайное подмножество?

**Варианты ответа:**

A. Чтобы удалить категориальные признаки.  
B. Чтобы сделать деревья менее похожими друг на друга и усилить пользу ансамбля.  
C. Чтобы каждое дерево стало линейной регрессией.  
D. Чтобы accuracy на train всегда была ниже baseline.  

**Правильный ответ:** B.


## 5. n_estimators и качество

**Тип:** выбор ответа

**Задание.** Увеличили `n_estimators` с 20 до 300. Validation-качество почти не изменилось, обучение стало дольше. Что это значит?

**Варианты ответа:**

A. Ансамбль мог уже стабилизироваться; дальнейшее увеличение числа деревьев даёт мало пользы за большее время.  
B. Больше деревьев всегда обязаны давать большой прирост.  
C. Модель сломалась: лес не может обучаться на 300 деревьях.  
D. Нужно обязательно уменьшить train до одного объекта.  

**Правильный ответ:** A.


## 6. Сравнение bagging и boosting

**Тип:** выбор ответа

**Задание.** Ученик говорит: “bagging и boosting одинаковы, потому что оба используют много деревьев”. Как уточнить?

**Варианты ответа:**

A. Bagging всегда использует линейные модели, boosting — только kNN.  
B. Boosting не является ансамблем.  
C. Разницы нет, отличаются только названия классов sklearn.  
D. Bagging обучает модели более независимо и снижает разброс, boosting строит модели последовательно, исправляя ошибки предыдущих.  

**Правильный ответ:** D.


## 7. class_weight и порог

**Тип:** выбор ответа

**Задание.** В случайном лесу для дисбалансной классификации включили `class_weight`, но recall важного класса всё ещё мал. Что можно проверить дальше?

**Варианты ответа:**

A. Только увеличить число строк в test.  
B. Удалить confusion matrix, чтобы не видеть ошибок.  
C. Подбор порога по `predict_proba` на validation и метрики precision/recall.  
D. Заменить все признаки на целевую переменную.  

**Правильный ответ:** C.


## 8. Impurity importance в лесу

**Тип:** выбор ответа

**Задание.** У признака с большим числом возможных значений высокая impurity importance. Почему стоит перепроверить это permutation importance?

**Варианты ответа:**

A. Permutation importance запрещена для RandomForest.  
B. Impurity importance иногда смещена в пользу признаков с большим числом вариантов разбиения.  
C. Если impurity importance высокая, проверка невозможна.  
D. Permutation importance измеряет только скорость обучения.  

**Правильный ответ:** B.


## 9. Лес и интерпретируемость

**Тип:** выбор ответа

**Задание.** Случайный лес дал качество выше одиночного дерева. Что обычно стало хуже по сравнению с одним деревом?

**Варианты ответа:**

A. Возможность получить предсказание.  
B. Способность работать с числовыми признаками.  
C. Возможность считать метрики.  
D. Интерпретируемость отдельного решения: вместо одного набора правил есть голосование многих деревьев.

## 10. Когда OOB ненадёжен

**Тип:** выбор ответа

**Задание.** В train очень мало объектов, а классы сильно несбалансированы. Почему OOB-оценка может быть шумной?

**Варианты ответа:**

A. Потому что OOB использует только объекты, не попавшие в bootstrap конкретных деревьев; при малых данных оценка может иметь большой разброс.  
B. Потому что OOB всегда считает качество на train без исключений.  
C. Потому что OOB отключает все деревья.  
D. Потому что OOB работает только в регрессии.  

**Правильный ответ:** A.
