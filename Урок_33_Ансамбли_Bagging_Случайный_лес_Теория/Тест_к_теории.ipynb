## 4. Усреднение прогнозов ансамбля

**Тип:** вычисление / код

**Задание.** Прогнозы трёх моделей `[10, 14, 12]`, истина `y=12`. Сравните средний квадрат ошибок одиночек и квадрат ошибки среднего прогноза.

**Правильный ответ:** MSE одиночных `8/3`, MSE среднего `0` (assert).


```python
import numpy as np
preds = np.array([10.0, 14.0, 12.0])
y = 12.0
errors = preds - y
mse_ind = float(np.mean(errors ** 2))
mse_ens = float((preds.mean() - y) ** 2)
assert abs(mse_ind - 8 / 3) < 1e-12
assert mse_ens == 0.0
print("ok", mse_ind, mse_ens)
```
