## 1. Идея bagging

**Тип:** выбор ответа

**Задание.** Зачем bagging обучает много моделей на разных bootstrap-выборках и усредняет их ответы?

**Варианты ответа:**

A. Чтобы увеличить разброс отдельных моделей.  
B. Чтобы запретить использовать train.  
C. Чтобы снизить variance ансамбля по сравнению с одной нестабильной моделью.  
D. Чтобы превратить классификацию в регрессию.

**Правильный ответ:** C.


## 2. Bootstrap

## 7. Глубина деревьев в лесу

**Тип:** открытый ответ

**Задание.** Почему случайный лес с очень глубокими деревьями может работать лучше одного очень глубокого дерева, но все равно не защищен от всех проблем переобучения?

**Формат ответа:** короткое текстовое объяснение.

**Правильный ответ:** усреднение многих деревьев снижает variance, но слишком сложные деревья и утечки в признаках все равно могут ухудшать обобщение.
