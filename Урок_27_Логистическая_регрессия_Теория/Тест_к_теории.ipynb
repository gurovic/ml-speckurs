## 1. Что предсказывает логистическая регрессия

**Тип:** выбор ответа

**Задание.** Почему логистическую регрессию используют для классификации, хотя в названии есть слово "регрессия"?

**Варианты ответа:**

A. Она напрямую предсказывает непрерывную цену объекта.  
B. Она работает только с задачами без целевой переменной.  
C. Она оценивает вероятность класса, а затем класс можно получить по порогу.  
D. Она всегда строит дерево решений.

**Правильный ответ:** C.

**Варианты ответа:**

A. Recall всегда становится равен нулю.  
B. Recall нельзя считать для положительного класса.  
C. Порог влияет только на train, но не на validation.

**Варианты ответа:**

## 8. Взвешивание классов
