## 1. Вероятность и класс — не одно и то же

**Тип:** выбор ответа

**Задание.** В модели оттока `predict_proba` выдал для клиента вероятность ухода `0.42`. В продукте менеджер хочет обзванивать только тех, у кого риск ухода не меньше `0.60`. Какой вывод корректен?

**Варианты ответа:**

A. Клиента нужно обзвонить, потому что `0.42` больше стандартного порога `0.5`.  
B. Модель обязана вернуть класс `1`, потому что вероятность не равна нулю.  
C. Клиента не нужно обзванивать при выбранном бизнес-пороге `0.60`, хотя модель всё равно считает риск ненулевым.  
D. `predict_proba` нельзя использовать для принятия решений, только `predict`.  

**Правильный ответ:** C.


## 2. Почему accuracy может обмануть

**Тип:** выбор ответа

**Задание.** В датасете 96% объектов относятся к классу `0`. Модель всегда предсказывает `0` и получает accuracy `0.96`. Почему это слабый результат для задачи поиска редкого класса `1`?

**Варианты ответа:**

A. Потому что высокая accuracy здесь может быть достигнута игнорированием редкого, но важного класса.  
B. Потому что accuracy всегда равна recall.  
C. Потому что accuracy нельзя считать для бинарной классификации.  
D. Потому что модель с accuracy выше `0.9` обязательно переобучена.  

**Правильный ответ:** A.


## 3. Порог зависит от цены ошибок

**Тип:** выбор ответа

**Задание.** Модель ищет опасные поломки оборудования. Пропуск настоящей поломки намного дороже ложной тревоги. Что разумнее сделать с порогом классификации, если модель выдаёт вероятности?

**Варианты ответа:**

A. Поднять порог, чтобы модель реже предупреждала о поломках.  
B. Оставить только accuracy и не смотреть confusion matrix.  
C. Выбирать порог по test после финальной отправки.  
D. Подобрать более низкий порог на validation, чтобы увеличить recall важного класса.  

**Правильный ответ:** D.


## 4. ROC-AUC и конкретный порог

**Тип:** выбор ответа

**Задание.** У двух моделей одинаковый ROC-AUC, но при рабочем пороге `0.7` одна даёт намного больше полезных срабатываний. Какой вывод наиболее точен?

**Варианты ответа:**

A. Если ROC-AUC одинаковый, confusion matrix при любом пороге тоже обязана совпасть.  
B. ROC-AUC описывает ранжирование по всем порогам, но не гарантирует лучший результат при конкретном рабочем пороге.  
C. ROC-AUC можно считать только для классов, а не для вероятностей.  
D. Значит, обе модели имеют одинаковые коэффициенты.  

**Правильный ответ:** B.


## 5. PR-кривая при редком положительном классе

**Тип:** выбор ответа

**Задание.** Положительный класс встречается редко, и важны именно найденные положительные объекты. Почему PR-кривая может быть полезнее ROC-кривой для обсуждения качества?

**Варианты ответа:**

A. PR-кривая не зависит от вероятностей модели.  
B. PR-кривая всегда показывает accuracy.  
C. Она фокусируется на precision и recall положительного класса, поэтому лучше показывает цену ложных срабатываний при редком классе.  
D. PR-кривая игнорирует false positive.  

**Правильный ответ:** C.


## 6. Регуляризация и коэффициенты

**Тип:** выбор ответа

**Задание.** После OneHotEncoding появились сотни признаков. Без регуляризации коэффициенты логистической регрессии стали огромными и нестабильными. Что делает L2-регуляризация?

**Варианты ответа:**

A. Добавляет штраф за большие коэффициенты и обычно делает решение устойчивее.  
B. Автоматически превращает все категориальные признаки обратно в строки.  
C. Гарантирует идеальный recall на train.  
D. Заменяет log loss на accuracy во время обучения.  

**Правильный ответ:** A.


## 7. Параметр C в LogisticRegression

**Тип:** выбор ответа

**Задание.** У `LogisticRegression` в sklearn увеличили `C` с `0.1` до `100`. Как это влияет на регуляризацию?

**Варианты ответа:**

A. Регуляризация становится сильнее, коэффициенты сильнее сжимаются.  
B. `C` меняет только случайное разбиение train/test.  
C. `C` используется только в линейной регрессии, не в логистической.  
D. Регуляризация становится слабее, модель получает больше свободы подстраиваться под данные.  

**Правильный ответ:** D.


## 8. class_weight не лечит всё

**Тип:** выбор ответа

**Задание.** При дисбалансе классов ученик поставил `class_weight=balanced` и решил, что теперь метрики можно не смотреть. Что не так?

**Варианты ответа:**

A. `class_weight` запрещён для логистической регрессии.  
B. Веса меняют обучение, но качество всё равно нужно проверять на validation подходящими метриками и порогом.  
C. После `class_weight` модель перестаёт выдавать вероятности.  
D. Балансировка классов всегда делает precision равным recall.  

**Правильный ответ:** B.


## 9. Утечка при кодировании категорий

**Тип:** выбор ответа

**Задание.** Категориальные признаки закодировали с учётом всей таблицы, включая test, а затем начали выбирать модель. Почему это рискованно?

**Варианты ответа:**

A. Кодирование категорий всегда ухудшает качество логистической регрессии.  
B. Логистическая регрессия принимает только текстовые признаки.  
C. Так в предобработку может просочиться информация из test; кодировщик нужно обучать внутри train/CV-процесса.  
D. Проблема возникает только при числовых признаках, а не при категориях.  

**Правильный ответ:** C.


## 10. Калибровка вероятностей

**Тип:** выбор ответа

**Задание.** Модель хорошо ранжирует клиентов: те, кому она даёт больший риск, чаще уходят. Но среди клиентов с прогнозом около `0.8` реально уходят только примерно 50%. Какой вывод корректен?

**Варианты ответа:**

A. Если ranking хороший, вероятности автоматически равны реальным частотам.  
B. Такое возможно только у решающих деревьев, но не у логистической регрессии.  
C. Нужно заменить все вероятности на классы и забыть о порогах.  
D. У модели может быть хороший ranking, но плохо откалиброванные вероятности.  

**Правильный ответ:** D.
